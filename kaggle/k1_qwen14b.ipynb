# K1: Qwen3-14B so với Qwen3-8B trên train (ZAIC DataMate, kế hoạch §11)

Cài đặt notebook: **Accelerator = GPU T4 x2**, **Internet = On**, Add Input = Dataset riêng tư chứa `zaic_code.zip` và `zaic_train.zip` (tạo bằng `python tools/kaggle_pack.py` trên laptop).

Chạy bằng **Save Version → Save & Run All** (chạy nền, không cần giữ trình duyệt). Xong thì tải `k1_results.zip` ở tab Output. Chỉ dữ liệu train; không đưa dữ liệu test lên Kaggle.

```python
import pathlib, shutil, zipfile
INP, WORK = pathlib.Path('/kaggle/input'), pathlib.Path('/kaggle/working')
ROOT = WORK / 'ZAIC'
if not (ROOT / 'datamate').exists():
    z = next(INP.rglob('zaic_code.zip'), None)
    if z:
        zipfile.ZipFile(z).extractall(ROOT)
    else:  # Kaggle đã tự giải nén file zip khi tạo Dataset
        src = next(p.parent.parent for p in INP.rglob('datamate/__init__.py'))
        shutil.copytree(src, ROOT, dirs_exist_ok=True)
train = ROOT / 'data' / 'train'
if not (train / 'questions.jsonl').exists():
    z = next(INP.rglob('zaic_train.zip'), None)
    if z:
        zipfile.ZipFile(z).extractall(ROOT / 'data')
    else:
        src = next(p.parent for p in INP.rglob('questions.jsonl') if (p.parent / 'rulebook.md').exists())
        shutil.copytree(src, train, dirs_exist_ok=True)
print(sorted(p.name for p in ROOT.iterdir()), (train / 'questions.jsonl').exists())
```
